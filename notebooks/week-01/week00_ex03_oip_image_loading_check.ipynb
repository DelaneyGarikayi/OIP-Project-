# Problem 0.3 — OIP image-loading check


This notebook checks that the small `oip` course library is importable and can load an image. The loader may use a course image if available, otherwise it falls back to a standard scikit-image sample.

```python
import numpy as np
import matplotlib.pyplot as plt

try:
    import oip
    oip_available = True
except Exception as exc:
    oip_available = False
    print('Could not import oip:')
    print(repr(exc))

print('oip available:', oip_available)
```

```python
assert oip_available is True, 'The oip package must be installed/importable in this environment.'
```

Load a grayscale image as floating point data and inspect its basic properties.

```python

image = oip.load_grey('camera', dtype=float)
image_shape = image.shape
image_dtype = image.dtype
image_min = float(image.min())
image_max = float(image.max())


print('shape:', image_shape)
print('dtype:', image_dtype)
print('range:', image_min, image_max)
```

```python
assert image.ndim == 2
assert np.issubdtype(image.dtype, np.floating)
assert 0.0 <= image_min <= image_max <= 1.0
assert image_shape[0] > 10 and image_shape[1] > 10
```

Display the image. This is mostly a visual check that plotting works in your notebook environment.

```python
plt.figure(figsize=(4, 4))
plt.imshow(image, cmap='gray', vmin=0, vmax=1)
plt.axis('off')
plt.title('OIP image-loading check');
```
