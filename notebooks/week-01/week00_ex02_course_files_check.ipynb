# Problem 0.2 — Course folder check


This notebook checks that you are running notebooks from a sensible course folder, not directly from a ZIP archive or a temporary download location.

```python
from pathlib import Path
import os

current_dir = Path.cwd()
print('Current working directory:')
print(current_dir)

# Does this path look like a temporary download directory?
path_text = str(current_dir).lower()
looks_temporary = any(word in path_text for word in ['tmp', 'temp'])
print('Looks temporary:', looks_temporary)
```

Choose the folder that you want to use as your course folder. This can be the current folder or a parent folder.

```python
# Replace the value below if needed.
# Use a relative path such as '.' or '..', or an absolute path.

course_dir = Path('.')

course_dir = Path(course_dir).expanduser().resolve()
print('Course directory:', course_dir)
print('Exists:', course_dir.exists())
```

Create a small file in the current folder to verify that you can write notebook outputs.
