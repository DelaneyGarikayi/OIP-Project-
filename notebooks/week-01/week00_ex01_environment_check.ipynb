The following cell will print the path of the current project root folder (the one that contains the folder `src/oip`).

```python
def find_project_src(package="oip"):
    """Find a nearby src/<package> directory and add src to sys.path."""
    cwd = Path.cwd().resolve()
    candidates = [cwd, *cwd.parents]
    for parent in list(candidates):
        try:
            candidates.extend(
                p for p in parent.iterdir()
                if p.is_dir()
            )
        except PermissionError:
            pass

    for root in candidates:
        src = root / "src"
        if (src / package).is_dir():
            sys.path.insert(0, str(src))
            return root

    raise RuntimeError(f"Could not find src/{package}")

PROJECT_ROOT = find_project_src("oip")
print(f"OIP project: {PROJECT_ROOT}")
```

## OIP package installation

The following cell will check, if the oip package is installed in the current environment. If not, it will print some instruction that match you current setup.

```python
from pathlib import Path
import shutil


try:
    import oip
    print('OIP is installed.')
except ModuleNotFoundError:
    # ------------------------------------------------------------
    # Determine current notebook path (CNP)
    # ------------------------------------------------------------
    try:
        import ipynbname

        CNP = ipynbname.path().resolve()
        search_start = CNP.parent

    except (ImportError, FileNotFoundError):
        # Fallback if the notebook filename cannot be determined.
        CNP = Path.cwd().resolve()
        search_start = CNP

    # ------------------------------------------------------------
    # Find project root containing src/oip
    # ------------------------------------------------------------
    CPP = None

    for parent in [search_start, *search_start.parents]:
        if (parent / "src" / "oip").is_dir():
            CPP = parent
            break

    # ------------------------------------------------------------
    # Give setup instructions
    # ------------------------------------------------------------
    if CPP is None:
        print(
            "ERROR: The 'oip' package is not installed and no "
            "'src/oip' directory could be found above the notebook location.\n\n"
            f"Current notebook location:\n    {CNP}"
        )

    elif shutil.which("uv") is not None:
        print(
            "The 'oip' package is not installed.\n\n"
            "uv was detected on this system.\n"
            "Open a terminal and run:\n\n"
            f"    cd {CPP}\n"
            "    uv sync\n\n"
            "Then select the project's .venv Python kernel "
            "and restart the notebook."
        )

    elif shutil.which("conda") is not None:
        print(
            "The 'oip' package is not installed.\n\n"
            "Conda was detected on this system.\n"
            "Open an Anaconda/Conda shell, activate the environment "
            "used by this notebook, and run:\n\n"
            f"    cd {CPP}\n"
            "    python -m pip install -e .\n\n"
            "Then restart the notebook kernel."
        )

    else:
        print(
            "ERROR: The 'oip' package is not installed and neither "
            "uv nor Conda could be detected.\n\n"
            f"Project directory:\n    {CPP}"
        )
```
