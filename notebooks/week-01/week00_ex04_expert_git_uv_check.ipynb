# Problem 0.4 — Expert git/uv check

This optional notebook is for the expert workflow. It checks that the notebook is running inside a git/uv-based course checkout. If you use the beginner Anaconda/ZIP workflow, you can skip this notebook unless instructed otherwise.

```python
from pathlib import Path
import shutil
import subprocess

def run(command):
    result = subprocess.run(command, shell=True, text=True, capture_output=True)
    print('$', command)
    print(result.stdout.strip())
    if result.stderr.strip():
        print(result.stderr.strip())
    print('return code:', result.returncode)
    return result

git_available = shutil.which('git') is not None
uv_available = shutil.which('uv') is not None
inside_git_repo = (run('git rev-parse --is-inside-work-tree') if git_available else None)

print('git available:', git_available)
print('uv available:', uv_available)
```

Run `git status` and inspect the output. A clean repository is not required, but the command should run in the expert workflow.

```python

git_status = run('git status --short') if git_available else None
git_status_ok = git_available and git_status.returncode == 0


print('git status ok:', git_status_ok)
```

```python
# This is intentionally ungraded by default because the beginner workflow does not require git.
assert isinstance(git_status_ok, bool)
```

Run a package import through `uv run` if `uv` is available.

```python

if uv_available:
    uv_import = run(
        "uv run python -c 'import numpy, skimage; print(\"imports ok\")'"
    )
    uv_import_ok = uv_import.returncode == 0
else:
    uv_import_ok = False


print("uv import ok:", uv_import_ok)
```

```python
# Optional workflow check. Turn this into a graded test only if the expert workflow is mandatory.
assert isinstance(uv_import_ok, bool)
```
